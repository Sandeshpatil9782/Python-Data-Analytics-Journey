### What is the Average Of Purchase Amount?

### What Is The Minimum And Maximum Purchase ?

```python
mx=df["Purchase_Amount"].max()

mn=df["Purchase_Amount"].min()

ran=df["Purchase_Amount"].max()-df["Purchase_Amount"].min()

print('Maximum:-',mx)

print('Minimum:-',mn)

print("Range:-",ran)
```

### On average, how much do customers differ from the average purchase?

### Inter Quartile Range (IQR)

```python
Q1=df["Purchase_Amount"].quantile(.25)

Q3=df["Purchase_Amount"].quantile(.75)

IQR=Q3-Q1

IQR
```

### Who are the top spending customers?

```python
p90 = df["Purchase_Amount"].quantile(0.90)

print(f"90th Percentile Purchase Amount: ₹{p90:.2f}")

top_customers = df[df["Purchase_Amount"] >= p90]

top_customers = top_customers[
    [
        "Customer_ID",
        "City",
        "Product_Category",
        "Purchase_Amount",
        
    ]
]

top_customers.sort_values(
    by="Purchase_Amount",
    ascending=False
    
).head(10).style.hide(axis="index")
```

```python
plt.figure(figsize=(8,3))

sns.violinplot(
    x=df["Purchase_Amount"],
    color="lightgreen"
)

plt.title("Purchase Amount Violin Plot")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.kdeplot(
    df["Discount_Percentage"],
    fill=True
)

plt.title("Discount Percentage Density Curve")

plt.show()
```

```python
#colo=['red','green','blue','yellow','orange']
sns.countplot(
    data=df,
    x="Product_Category",
    hue='Product_Category',
    palette='Set1'
)

plt.title("Orders by Product Category")

plt.xticks(rotation=20)

plt.show()
```

```python
payment = df["Payment_Method"].value_counts()

plt.figure(figsize=(4,5))

plt.pie(
    payment,
    labels=payment.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Method Distribution")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Purchase_Amount"
)

plt.title("Age vs Purchase Amount")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.ecdfplot(df["Purchase_Amount"])

plt.title("Empirical CDF")

plt.show()
```

```python
plt.figure(figsize=(6,6))

sts.probplot(
    df["Purchase_Amount"],
    dist="norm",
    plot=plt
)

plt.title("QQ Plot")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(
    df.select_dtypes("number").corr(),
    annot=True,
    cmap="viridis"
)

plt.title("Correlation Matrix")

plt.show()
```
