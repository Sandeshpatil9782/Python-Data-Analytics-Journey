# Inferential Statistics
    Inferential Statistics is a branch of statistics that uses data collected from a sample to make conclusions, predictions, or decisions about an entire population.

    Unlike Descriptive Statistics, which summarizes the avalable data, Inferential Statistics helps us generalize the findings from a sample to a larger population.

    Inferential Statistics is widely used in business analytics, healthcare, finance, marketing, scientific research, and machine learning because collecting data from every individual in a population is often expensive, time-consuming, or impossible.

### Descriptive Statistics vs Inferential Statistics

| Descriptive Statistics | Inferential Statistics |
|-------------------------|------------------------|
| Summarizes existing data | Makes conclusions about a population |
| Works only with available data | Uses sample data to estimate population characteristics |
| No prediction is made | Used for prediction and decision making |
| Examples: Mean, Median, Standard Deviation | Examples: Confidence Interval, Hypothesis Testing, t-Test |

### Population vs Sample

| Population | Sample |
|------------|---------|
| Complete collection of observations | Subset of the population |
| Usually very large | Smaller than the population |
| Represented by N | Represented by n |
| Expensive and time-consuming to collect | Faster and less expensive |
| Used in a census | Used in sampling |

### Census
    A Census is the process of collecting information from every member of a population.

    Since every observation is included, census results are highly accurate.

    However, conducting a census is often expensive, time-consuming, and impractical for large populations.

## Sampling
    Sampling is the process of selecting a subset of observations from a population for analysis.

    Instead of studying the entire population, only a representative sample is analyzed.

    The conclusions obtained from the sample are then generalized to the population.

    The Mean of population is related or approximately Within the Mean Calculated By Sample.(We Can Give Find a Range that Means fit.)

### Limitations of Sampling
    - Results depend on sample quality.
    - A poor sample may produce biased results.
    - Sampling error may occur.
    - Incorrect sampling methods can lead to inaccurate conclusions.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import scipy.stats as st
```

### Parameter
    A Parameter is a numerical value that describes a characteristic of an entire population.

    Since a parameter is calculated using every observation in the population, it is considered a fixed value.

    Examples of parameters include:
    - Population Mean
    - Population Variance
    - Population Standard Deviation
    - Population Proportion

### Statistic
    A Statistic is a numerical value calculated from a sample.

    Statistics are used to estimate the corresponding population parameters.

    Examples of statistics include:
        - Sample Mean
        - Sample Variance
        - Sample Standard Deviation
        - Sample Proportion

```python
population_mean = Employees["Monthly_Salary"].mean()

sample_mean = sample_100["Monthly_Salary"].mean()

print(f"Population Mean (μ): {population_mean:.2f}")

print(f"Sample Mean (x̄): {sample_mean:.2f}")
```

```python
population_std = Employees["Monthly_Salary"].std(ddof=0)

sample_std = sample_100["Monthly_Salary"].std()

print(f"Population Standard Deviation (σ): {population_std:.2f}")

print(f"Sample Standard Deviation (s): {sample_std:.2f}")
```

### Sampling Unit
    A Sampling Unit is a single member or observation selected from the population.
    
    Example:
        One employee selected for the survey.

### Types of Sampling
    Sampling techniques are the methods used to select observations from a population.
    
        1. Probability Sampling
        2. Non-Probability Sampling

### 1.Probability Sampling
    Probability Sampling is a sampling technique in which every member of the population has a known and non-zero chance of being selected.

    Types of Non-Probability Sampling
        - Sample Random Sampling
        - Systematic Sampling
        - Stratified Sampling
        - Cluster Sampling

### 1.1 Simple Random Sampling(SRS)

    Simple Random Sampling is a technique in which every observation in the population has an equal probability of being selected.


```python
random_sample = Employees.sample(n=100,random_state=42)

random_sample.head()
```

### 1.2 Systematic Sampling
    Systematic Sampling selects observations at a fixed interval.

```python
k = 10

systematic_sample = Employees.iloc[::k]

systematic_sample.head()

```

### 1.3 Stratified Sampling
    Stratified Sampling divides the population into homogeneous groups called strata.



```python
stratified_sample = (
            Employees
            .groupby("Department",group_keys=False)
            .sample(n=20,random_state=42)
    )

stratified_sample.head()
```

### 1.4 Cluster Sampling

    Cluster Sampling divides the population into naturally occurring groups called clusters.


```python
Employees["Location"] = np.random.choice( [ "Bengaluru","Mumbai","Delhi","Pune"   ],

    len(Employees)
                                        )

selected_clusters = ["Bengaluru","Pune"]

cluster_sample = Employees[

    Employees["Location"].isin(selected_clusters)
]

cluster_sample.head()



                                        
```

### 2. Non-Probability Sampling
    Non-Probability Sampling is a sampling technique in which not every member of the population has a known or equal chance of being selected.

    Types of Non-Probability Sampling
        - Convenience Sampling
        - Judgment (Purposive) Sampling
        - Quota Sampling
        - Snowball Sampling

### 2.1 Convenience Sampling

    Convenience Sampling selects observations that are easiest to access.


```python
convenience_sample = Employees.head(100)

convenience_sample.head()
```

### 2.2 Judgment Sampling

    Judgment Sampling is a technique in which the researcher selects observations based on knowledge, experience, or specific criteria.


```python
Judg_sample=Employees[Employees['Experience'] >=15]

Judg_sample.head()
```

### 2.3 Quota Sampling
    Quota Sampling divides the population into groups and selects a fixed number of observations from each group.

    Unlike Stratified Sampling, the selection inside each group is not random.

```python
quota_sample = (
    Employees
    .groupby("Department",group_keys=False).head(30)

)

quota_sample.head()
```

```python
print(quota_sample['Department'].value_counts())
print('*'*25)
print('Sample Size(n)',len(quota_sample))
```

### 2.4 Snowball Sampling
    Snowball Sampling is used when the target population is difficult to identify.

    Existing participants help identify or recruit additional participants.

    The sample gradually grows like a snowball.

```python
sample = Employees.sample( n=250,random_state=42 )

comparison = pd.DataFrame({

    "Population":Employees["Monthly_Salary"],

    "Sample":sample["Monthly_Salary"].reindex(
        Employees.index
    )

})

plt.figure(figsize=(6,5))

sns.boxplot(data=comparison)

plt.title("Population vs Sample")

plt.show()
```

```python
population_mean = Employees["Monthly_Salary"].mean()

sample_mean = sample_100["Monthly_Salary"].mean()

plt.figure(figsize=(10,5))

sns.kdeplot(
    Employees["Monthly_Salary"],
    linewidth=2,
    label="Population",
    fill=True,
    alpha=0.3
)

sns.kdeplot(
    sample_100["Monthly_Salary"],
    linewidth=2,
    label="Sample",
    fill=True,
    alpha=0.3
)

plt.axvline(
    population_mean,
    color="blue",
    linestyle="--",
    linewidth=2,
    label=f"Pop Mean = {population_mean:.1f}"
)

plt.axvline(
    sample_mean,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Samp Mean = {sample_mean:.1f}"
)

plt.title("Population vs Sample Distribution")

plt.xlabel("Monthly Salary")

plt.ylabel("Density")

plt.legend(loc=2)

plt.grid(alpha=0.3)

plt.show()
```
