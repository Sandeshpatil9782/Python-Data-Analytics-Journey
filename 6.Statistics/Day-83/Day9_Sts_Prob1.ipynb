# Probability & Basic Concept

## 🎲 Random Variable

    A Random Variable (RV) is a variable whose value depends on the outcome of a random experiment.

    Instead of knowing the exact result beforehand, the value is determined after the experiment is performed.

    A random variable is usually represented by X, Y, or Z.

### Example

    Suppose an online food delivery company randomly selects one customer order.

    The value of X (Delivery Time) is not known until the order is delivered.

    Therefore,

        X = Delivery Time (minutes) is a Random Variable.

### Types Of Random Variable

### 1. Discrete Random Variable

    A discrete random variable can take countable values.

    Examples:
        - Number of Orders
        - Number of Cancelled Orders
        - Number of Customers
        - Number of Returned Orders

    Example Values:
    0, 1, 2, 3, 4, ...

### 2. Continuous Random Variable

    A continuous random variable can take any value within a range.

    Examples:
    - Delivery Time (minutes)
    - Order Value (₹)
    - Customer Weight
    - Customer Height

    Example Values:
        18.5, 20.3, 45.7, 75.2, ...

### Probability
    Probability is a branch of statistics that measures the likelihood or chance of an event occurring.

    It helps us quantify uncertainty and predict possible outcomes based on historical data or experiments.

    ### Probability Range

    - 0 → Impossible Event
    - 1 → Certain Event
    - 0 < P(E) < 1 → Possible Event

    Probability is widely used in data analysis, finance, healthcare, banking, marketing, insurance, and machine learning to make informed decisions under uncertainty.

## Why Do We Need Probability?

    Probability helps organizations make decisions when outcomes are uncertain.

### Business Applications

    - Predict customer purchases.
    - Estimate loan default risk.
    - Forecast product demand.
    - Detect fraudulent transactions.
    - Analyze customer behavior.
    - Support machine learning models.

### Real-World Example

    An online food delivery company wants to estimate the probability that a randomly selected order will be delivered successfully.

## Basic Terminologies

### 1. Experiment
    An action or process that produces one or more outcomes.

    Example: Selecting one order randomly.
---
### 2. Outcome
    A single possible result of an experiment.
    
    Example: Delivered
---
### 3. Sample Space (S)

    The set of all possible outcomes.

    Example:
        S = {Delivered, Cancelled, Returned}
---

### 4. Event (E)
    An event is one or more outcomes of interest from the sample space.

    Example:
        E = Order is Delivered

##  Probability Formula

    The probability of an event is calculated by dividing the number of favorable outcomes by the total number of possible outcomes.

$$
P(E)=\frac{\text{Number of Favorable Outcomes}}
{\text{Total Number of Outcomes}}
$$

    Where:
        - P(E) = Probability of Event E
        - Favorable Outcomes = Desired outcomes
        - Total Outcomes = Total possible outcomes

###  Characteristics of Probability
     - Probability is always between '0 and 1'.
    - A probability of '0' means the event cannot occur.
    - A probability of '1' means the event will definitely occur.
    - The higher the probability, the more likely the event is to happen.
    - The lower the probability, the less likely the event is to happen.

### Business Problem

    An online food delivery company wants to analyze customer orders.

    As a Data Analyst, your task is to estimate the probability of different events, such as:

- An order being delivered.
- A customer choosing UPI.
- An order being placed on the weekend.

These insights help management understand customer behavior and improve business decisions.

### Probability Of Delivered Orders

```python
total_orders = len(Orders)

delivered = len(
    Orders[
        Orders["Order_Status"]=="Delivered"
    ]
)

prob = delivered/total_orders

print("Total Orders :",total_orders)

print("Delivered :",delivered)

print("Probability :",round(prob,2))
```

### Probability Of Weekend Orders

```python
weekend = len(
    Orders[
        Orders["Weekend_Order"]=="Yes"
    ]
)

prob_weekend = weekend/total_orders

print("Weekend Orders :",weekend)

print("Probability :",round(prob_weekend,2))
```

### Probability of Payment Method

    The probability of each payment method is calculated by dividing the number of orders using that payment method by the total number of orders.

    Formula:
        P(Payment Method) = Number of Orders Using That Payment Method / Total Number of Orders

    The `value_counts(normalize=True)` function automatically performs this calculation and returns the probability of each category.

```python
Orders["Payment_Method"].value_counts(normalize=True)
```
