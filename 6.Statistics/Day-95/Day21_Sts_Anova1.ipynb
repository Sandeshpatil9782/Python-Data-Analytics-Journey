# ANOVA (Analysis of Variance)

    ANOVA (Analysis of Variance) is a statistical hypothesis test used to compare the means of three or more independent groups.

    Instead of performing multiple independent t-tests, ANOVA determines whether at least one group mean is significantly different from the others using a single statistical test.

    Also Called As F-Test Or F-Statistics

# Types Of Anova

## 1. One-Way ANOVA
    One-Way ANOVA is used when there is one independent variable (factor) with multiple groups and one continuous dependent variable.

    Use One-Way ANOVA when
        • Comparing three or more independent groups
        • Dependent variable is numerical
        • One categorical independent variable.

    Examples
        • Compare average salaries across departments.
        • Compare customer satisfaction among three stores.
 

### Hypotheses
    Null Hypothesis:- All population means are equal.
$$
H0​:μ1​=μ2​=μ3​=⋯=μk​
$$
    
    Alternative Hypothesis:- At least one population mean is different.
$$
H_1:At\ least\ one\ population\ mean\ differs
$$

### Assumptions of One-Way ANOVA
    1. Independence
            Each observation should be independent of the others.

        Example
            Sales of one employee should not influence another employee's sales.

    2. Normality
        The dependent variable should be approximately normally distributed within each group.

    This assumption can be checked using
        • Shapiro-Wilk Test
        • Q-Q Plot
        • Histogram

    3. Homogeneity of Variance
        The population variances of all groups should be approximately equal.
    
    This assumption can be checked using

        • Levene's Test
        • Bartlett's Test

### Between Group Variation
    Between-group variation measures how far each group mean is from the overall mean.

    Large between-group variation indicates that the group means are different from each other.

    Example
        Training A Average Sales = 52
        Training B Average Sales = 70
        Training C Average Sales = 88

            These means are far apart, so between-group variation is high.

### Within Group Variation
    Within-group variation measures how much the observations vary inside each group.

    Small within-group variation indicates that observations inside a group are similar.

    Example
        Training A
            50 51 53 52 54

    The values are close together, so within-group variation is small.

### F Statistic
    The F-statistic is the test statistic used in One-Way ANOVA.

    It compares the variation between groups with the variation within groups.

    Formula

$$
F=\frac{\text{Mean Square Between Groups (MSB)}}{\text{Mean Square Within Groups (MSW)}}
$$

    or

$$
F=\frac{\text{Between Group Variance}}{\text{Within Group Variance}}
$$

    Interpretation
        • If the F-statistic is close to 1, the between-group variation is similar to the within-group variation. This suggests that the group means are approximately equal.

        • If the F-statistic is much greater than 1, the between-group variation is much larger than the within-group variation. This indicates that at least one group mean is significantly different.

    In simple words,
    - Large F Statistic → Reject the Null Hypothesis
    - Small F Statistic → Fail to Reject the Null Hypothesis

### Total Sum of Squares (SST)

    The Total Sum of Squares measures the total variation of all observations from the overall (grand) mean.

    Formula

$$
SST=\sum_{i=1}^{N}(X_i-\bar{X})^2
$$

    where

- $X_i$ = Individual observation
- $\bar{X}$ = Grand Mean
- $N$= Total number of observations

### Sum of Squares Between Groups (SSB)
    SSB measures how much the group means differ from the overall mean.

    Formula
$$
SSB=\sum_{i=1}^{k}n_i(\bar{X_i}-\bar{X})^2
$$

    where

- $k$ = Number of groups
- $n_i$ = Number of observations in group \(i\)
- $\bar{X_i}$ = Mean of group \(i\)
- $\bar{X}$ = Grand Mean



### Sum of Squares Within Groups (SSW)
    SSW measures the variation of observations around their own group mean.

    Formula
$$
SSW=\sum_{i=1}^{k}\sum_{j=1}^{n_i}(X_{ij}-\bar{X_i})^2
$$

    A smaller SSW indicates that observations inside each group are more similar.

### fundamental equation of One-Way ANOVA
$$
SST=SSB+SSW
$$

### Degrees of Freedom
    Between Groups
$$
df_{Between}=k-1
$$

    Within Groups
$$
df_{Within}=N-k
$$

    Total
$$
df_{Total}=N-1
$$

    where
- \(k\) = Number of groups
- \(N\) = Total observations

### Mean Squares
    Mean Square Between Groups
$$
MSB=\frac{SSB}{df_{Between}}
$$

    Mean Square Within Groups
$$
MSW=\frac{SSW}{df_{Within}}
$$

    Mean Squares are simply the average variation.

### Business Problem
    A company provides three different training programs to its employees.

    Management wants to determine whether the average monthly sales differ among the three training programs.

    Training A:- 55 60 58 62 57
    Training B:- 68 70 72 69 71
    Training C:- 80 82 78 81 79

    Significance Level
$$
\alpha=0.05
$$

### 
    Null Hypothesis
$$
H_0:\mu_A=\mu_B=\mu_C
$$
    
    Alternative Hypothesis
$$
H_1:\text{At least one group mean differs}
$$

```python
import numpy as np
import pandas as pd
```

```python

training = pd.DataFrame({

    "Training_A":[55,60,58,62,57],

    "Training_B":[68,70,72,69,71],

    "Training_C":[80,82,78,81,79]

})

training
```

### Group Mean

```python
training.mean()
```

### Grand Mean

```python
grand_mean = training.stack().mean()

print("Grand Mean :", round(grand_mean,2))
```

```python
# One Way Anova
from scipy.stats import f_oneway

F_statistic, p_value = f_oneway(

    training["Training_A"],

    training["Training_B"],

    training["Training_C"]

)

print("F Statistic :", round(F_statistic,4))

print("P Value :", round(p_value,6))
```

### Calculate Degrees of Freedom
    Between Groups (Numerator df)
$$
df_1=k−1
$$
    
        where:
            k = Number of groups

    Within Groups (Denominator df)
$$
df_2=N−k
$$
        
        where:
            N = Total number of observations.

    - 3 Training Programs
    - 5 employees in each group
        
        Then:- k=3 And N=15
$$
df_1=3−1=2
$$
	
$$
df_2=15−3=12
$$


```python
# Critical Value At Alpha=0.05

from scipy.stats import f

alpha = 0.05

df1 = 2      # Between Groups

df2 = 12     # Within Groups

critical = f.ppf(1-alpha, df1, df2)

print("Critical Value :", round(critical,4))
```

### Decision Rule
    Critical Value Method
        Reject the Null Hypothesis if
$$
F_{Calculated}>F_{Critical}
$$

        Otherwise,
            Fail to Reject the Null Hypothesis.

---

    P-value Method
        Reject the Null Hypothesis if
$$
p\text{-value}\le\alpha
$$

        Otherwise,
            Fail to Reject the Null Hypothesis.

```python
# Using F_Stats

if F_statistic > critical:

    print("Reject the Null Hypothesis")

else:

    print("Fail to Reject the Null Hypothesis")
```

```python
# Using p_value

alpha = 0.05

if p_value <= alpha:

    print("Reject the Null Hypothesis")

else:

    print("Fail to Reject the Null Hypothesis")
```
